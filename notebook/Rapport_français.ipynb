## 1. Introduction

La consommation d’électricité constitue un enjeu central pour les économies modernes, car elle reflète à la fois les comportements des ménages et l’activité économique, tout en posant des défis majeurs en matière de planification énergétique et de stabilité des réseaux. Dans un contexte de changement climatique et de variabilité météorologique croissante, comprendre les déterminants de la demande d’électricité est devenu particulièrement important.

Parmi ces déterminants, la température joue un rôle clé. Dans les pays où le chauffage repose en partie sur l’électricité, des températures plus froides sont généralement associées à une demande d’électricité plus élevée, tandis que les périodes plus chaudes correspondent à une consommation plus faible. Ces effets sont souvent saisonniers et peuvent varier selon les régions en fonction des conditions climatiques, de la densité de population et de la structure économique.

Ce projet analyse la relation entre la **température quotidienne et la consommation d’électricité dans les régions françaises** sur la période **2020–2024**. La France constitue un cas d’étude pertinent en raison de la diversité de ses climats régionaux, de la part relativement élevée du chauffage électrique (66 % selon EDF) et de la disponibilité de données ouvertes détaillées sur la consommation d’électricité et les conditions météorologiques.

La question de recherche principale est la suivante :

> **Dans quelle mesure la température quotidienne influence-t-elle la consommation d’électricité dans les régions françaises ?**

Pour répondre à cette question, l’analyse se déroule en trois étapes. Tout d’abord, des statistiques descriptives sont utilisées afin de documenter l’hétérogénéité régionale, la variabilité et les dynamiques saisonnières de la consommation d’électricité et de la température. Ensuite, la relation entre la température et la consommation d’électricité est explorée à l’aide de corrélations et d’analyses graphiques. Enfin, ces résultats descriptifs motivent une analyse économétrique plus formelle visant à quantifier l’association entre la température et la demande d’électricité, tout en tenant compte des différences régionales persistantes.

## 2. Sources 
Cette étude s'appuie sur deux sources principales de données ouvertes, toutes deux disponibles à une fréquence quotidienne et couvrant la période du **1er janvier 2020 au 31 décembre 2024**.

### 2.1 Données de consommation d'électricité

Les données de consommation d'électricité proviennent du gestionnaire du réseau de transport d'électricité français (RTE). Le jeu de données fournit **la consommation d'électricité quotidienne agrégée au niveau régional** pour la France métropolitaine, mesurée en mégawattheures (MWh).

Ces données permettent d'analyser les dynamiques temporelles et l'hétérogénéité régionale de la demande d'électricité à travers la France.

### 2.2 Données météorologiques

Les données météorologiques sont extraites de **l'API NASA POWER**, qui fournit des variables météorologiques standardisées issues d'observations satellites et de modèles de réanalyse. La variable principale utilisée dans cette étude est la **température moyenne quotidienne à 2 mètres (T2M)**, exprimée en degrés Celsius.

L'API permet un accès reproductible aux données météorologiques historiques pour des coordonnées géographiques spécifiques, rendant possible la construction d'indicateurs de température régionaux cohérents.

## 3. Préparation des données

Les données de consommation d’électricité et les données météorologiques proviennent de sources différentes et nécessitent donc un prétraitement avant toute analyse conjointe.

Les données de consommation d’électricité issues de RTE sont agrégées au **niveau régional quotidien**. Cette agrégation permet d’aligner les données sur la fréquence quotidienne des observations météorologiques et de lisser les fluctuations intra-journalières de très court terme, afin de se concentrer sur des variations de la demande plausiblement liées à la température.

Les données météorologiques sont construites à partir de **plusieurs points géographiques représentatifs au sein de chaque région**. Pour chaque région, la température moyenne quotidienne (T2M) est calculée comme la moyenne sur l’ensemble de ces points, ce qui permet de capturer les conditions climatiques régionales tout en réduisant la sensibilité à un lieu particulier.

## 5. Relation entre la température et la consommation d’électricité

Cette section examine la relation entre la température quotidienne et la consommation d’électricité en amont de l’analyse par régression. L’objectif est d’évaluer si la température est systématiquement associée à la demande d’électricité et de caractériser la nature de cette relation à l’aide de corrélations et d’analyses graphiques.

### Analyse des corrélations

Dans un premier temps, nous calculons la corrélation entre la température moyenne quotidienne (T2M) et la consommation d’électricité séparément pour chaque région.

L’intensité de la corrélation varie selon les régions, reflétant des différences de conditions climatiques, de besoins en chauffage et de niveaux de consommation de base. Toutefois, les corrélations ne permettent pas de distinguer la **variation intra-régionale dans le temps** des **différences de niveaux de consommation entre régions**. </center>

La consommation d’électricité en hiver dépasse celle de l’été dans toutes les régions, confirmant l’existence d’un effet saisonnier fort et systématique.

En moyenne, la consommation d’électricité en hiver est supérieure de 30 % à 60 % à celle de l’été selon les régions, avec des ratios hiver–été allant d’environ 1,3 dans les régions du sud comme PACA à plus de 1,6 en Île-de-France, ce qui indique une hétérogénéité marquée de la sensibilité saisonnière.

Il est notable que l’Île-de-France présente l’un des écarts hiver–été les plus élevés, bien qu’elle ne figure pas parmi les régions les plus froides. Cela suggère que la température explique principalement les fluctuations saisonnières de court terme au sein des régions, tandis que les différences persistantes de niveaux de consommation d’électricité sont davantage déterminées par des facteurs structurels que par le climat seul.

## 6. Analyse par régression : température et consommation d’électricité

Cette section quantifie la relation entre la température quotidienne et la consommation d’électricité à l’aide de modèles de régression linéaire. L’objectif est de mesurer la réponse de la demande d’électricité aux variations de température, tout en tenant compte des différences régionales persistantes dans les niveaux de consommation.

Tous les modèles sont estimés par la méthode des moindres carrés ordinaires (OLS). Des erreurs standards robustes à l’hétéroscédasticité (HC1) sont reportées afin de tenir compte d’une éventuelle hétéroscédasticité de la consommation quotidienne d’électricité.

### Régression OLS poolée

Nous estimons dans un premier temps une régression OLS poolée reliant la consommation d’électricité à la température moyenne quotidienne sur l’ensemble des régions et des dates :
$$
\text{Consommation}_{it} = \alpha + \beta \,\text{T2M}_{it} + \varepsilon_{it}
$$

où $\text{Consommation}_{it}$ désigne la consommation quotidienne d’électricité dans la région $i$ au jour $t$, et  
$\text{T2M}_{it}$ correspond à la température moyenne quotidienne.

Cette spécification capte l’association moyenne entre la température et la consommation d’électricité, sans tenir compte de l’hétérogénéité régionale.

<center>
  <img src="../figure/pool_OLS.png" alt="Résultats de la régression OLS poolée" style="width: 600px;"/>
</center>

Le coefficient estimé de la température est négatif et statistiquement significatif. Une augmentation d’un degré Celsius de la température moyenne quotidienne est associée à une diminution d’environ 5 160 MWh de la consommation d’électricité. Ce résultat est cohérent avec une demande plus élevée lors des journées froides en raison des besoins de chauffage.

Cependant, le modèle n’explique qu’une part limitée de la variation totale de la consommation d’électricité, avec un $R^2$ d’environ 0,13. Cela indique que la température seule ne permet pas d’expliquer les fortes différences de consommation entre régions.

### Modèle à effets fixes régionaux

Afin de contrôler les différences régionales persistantes, nous estimons un modèle incluant des effets fixes régionaux :
$$
\text{Consommation}_{it} = \alpha + \beta\,\text{T2M}_{it} + \gamma_i + \varepsilon_{it}
$$

où $\gamma_i$ capte les caractéristiques régionales invariantes dans le temps, telles que la taille de la population, l’activité économique ou le parc de logements.

<center>
  <img src="../figure/fixed_OLS.png" alt="Résultats du modèle à effets fixes régionaux" style="width: 650px;"/>
</center>

Une fois les effets fixes régionaux inclus, le coefficient de la température reste négatif, statistiquement significatif et de magnitude similaire (environ −5 070 MWh par °C). Cela suggère que la relation entre température et consommation reflète principalement des variations intra-régionales dans le temps, plutôt que des différences entre régions.

Le pouvoir explicatif du modèle augmente fortement, avec un $R^2$ d’environ 0,90, indiquant que l’hétérogénéité régionale explique l’essentiel de la variation de la consommation d’électricité. La température explique principalement les fluctuations de court terme au sein des régions.

Les effets fixes estimés confirment que l’Île-de-France présente une consommation d’électricité systématiquement plus élevée que les autres régions, même après contrôle de la température.

### Comparaison des spécifications

<center>
  <img src="../figure/compare.png" alt="Comparaison des modèles" style="width: 450px;"/>
</center>

La comparaison des deux spécifications montre que l’effet estimé de la température est stable d’un modèle à l’autre, tandis que l’introduction des effets fixes régionaux améliore fortement l’ajustement du modèle. Cela met en évidence le rôle dominant des facteurs structurels régionaux dans la détermination des niveaux de consommation, la température agissant comme un déterminant clé des variations de court terme.

La spécification à effets fixes isole la variation intra-régionale dans le temps, ce qui permet d’interpréter le coefficient de la température comme une réponse de court terme. Sans effets fixes, l’effet de la température est confondu avec la taille de la population, l’activité économique, le parc de logements, les modes de vie, etc. Il devient alors impossible de distinguer si une consommation plus élevée provient du climat ou simplement du fait qu’une région est plus grande.

Ces résultats doivent être interprétés comme des associations statistiques plutôt que comme des effets causaux. Néanmoins, ils sont cohérents avec l’analyse descriptive et fournissent des éléments solides indiquant que la température joue un rôle important dans la détermination des profils quotidiens de consommation d’électricité en France.

## 7. Discussion

Trois résultats principaux se dégagent de l’analyse.

Premièrement, la température est fortement et systématiquement associée à la consommation d’électricité : les journées plus froides coïncident avec une demande plus élevée dans toutes les régions. Le signe et l’ordre de grandeur de cette relation sont stables, tant dans les analyses descriptives que dans les estimations par régression.

Deuxièmement, l’hétérogénéité régionale domine largement les niveaux de consommation. Même après prise en compte de la température, les régions diffèrent fortement en termes de demande de base, ce qui est cohérent avec l’existence de différences structurelles persistantes dans le temps.

Troisièmement, l’Île-de-France illustre clairement cette distinction. Bien qu’elle ne fasse pas partie des régions les plus froides, elle demeure la région à la consommation la plus élevée sur l’ensemble des années et des saisons. Cela renforce l’interprétation selon laquelle la température explique principalement les variations de court terme, tandis que le niveau de consommation reflète des facteurs structurels qui ne sont pas capturés par les conditions météorologiques seules.

## 8. Limites

Cette étude met en évidence des associations statistiques robustes, mais plusieurs limites doivent être prises en compte dans l’interprétation des résultats.

- **Absence de causalité :** les coefficients de régression doivent être interprétés comme des associations conditionnelles et non comme des effets causaux. La température n’est pas assignée aléatoirement et la consommation d’électricité peut réagir à d’autres facteurs évoluant simultanément.

- **Variables omises :** les modèles se concentrent sur la température et les effets fixes régionaux. D’autres déterminants potentiels de la demande (activité économique, structure démographique, caractéristiques du bâti, efficacité énergétique, politiques tarifaires) ne sont pas inclus, en raison des contraintes de temps, de ressources et du périmètre du projet.

- **Choix d’agrégation :** la consommation d’électricité et les variables météorologiques sont agrégées au niveau régional quotidien. Cette agrégation peut masquer des variations intra-journalières ainsi qu’une hétérogénéité locale au sein des régions. De plus, l’utilisation de points météorologiques représentatifs peut ne pas capter pleinement les différences microclimatiques, en particulier dans les régions de grande taille.

Malgré ces limites, les résultats offrent une image cohérente et globalement consistante de la relation entre la température quotidienne et la consommation d’électricité régionale.

## 9. Conclusion

Ce projet propose une analyse empirique au niveau régional de la relation entre les variations quotidiennes de température et la demande d’électricité en France, en distinguant les effets de court terme liés à la température des différences régionales persistantes, pour 12 régions métropolitaines sur la période 2020–2024.

Les résultats mettent en évidence une relation négative claire et robuste : la demande d’électricité augmente lors des journées plus froides. Cette relation apparaît dans les corrélations et les analyses graphiques, et est confirmée par les régressions OLS. Lorsque des effets fixes régionaux sont introduits, l’effet de la température reste d’une ampleur similaire, ce qui soutient une interprétation fondée sur les variations intra-régionales dans le temps.

Parallèlement, la majeure partie des différences de niveaux de consommation d’électricité s’explique par une hétérogénéité régionale persistante. L’Île-de-France demeure un cas atypique de forte demande, même après contrôle de la température, soulignant l’importance de facteurs structurels allant au-delà du climat.

Ces résultats doivent être interprétés comme descriptifs et associatifs plutôt que causaux, compte tenu de l’absence de variation exogène de la température et de contrôles supplémentaires.

L’intégration (au-delà du champ de ce rapport) d’autres variables climatiques, telles que les degrés-jours de chauffage ou des indicateurs de températures extrêmes, de contrôles socio-économiques, ou d’effets fixes temporels, pourrait améliorer le pouvoir explicatif du modèle. L’exploration d’effets non linéaires de la température ou la distinction entre consommation résidentielle et industrielle permettraient également d’approfondir l’analyse des dynamiques de la demande d’électricité.

## 10. Modèles prédictifs

Jusqu’à présent, nous avons utilisé des méthodes statistiques et économétriques pour interpréter la relation entre la **température** et la **consommation d’électricité**, ce qui permet déjà de répondre à la question de recherche principale. Dans cette section, nous adoptons une approche différente en nous plaçant dans un cadre de **prévision**, qui constitue une extension du projet :

> **Étant donné des informations météorologiques et calendaires, dans quelle mesure peut-on prédire la demande d’électricité du lendemain dans chaque région ?**

### Méthode

La variable cible est la **consommation quotidienne d’électricité régionale**, mesurée en **MWh**.  
Les variables explicatives incluent :

- **Variables météorologiques** : indicateurs de température, humidité, rayonnement solaire, humidité du sol, pression atmosphérique, vent, précipitations.  
- **Effets calendaires** : mois, saison, indicateur de week-end.  
- **Identifiant régional** : variable catégorielle *Région*.

La séparation entre l’échantillon d’entraînement et l’échantillon de test respecte la structure temporelle des données :
- **Ensemble d’entraînement** : 2020–2023  
- **Ensemble de test** : 2024  

Les performances des modèles sont évaluées sur l’année 2024 à l’aide des indicateurs suivants :
- **RMSE** (Root Mean Squared Error),  
- **MAE** (Mean Absolute Error),  
- **R²** (coefficient de détermination).

### Sélection des variables

L’utilisation de l’ensemble des variables météorologiques disponibles augmenterait la complexité du modèle et introduirait une redondance importante en raison de fortes corrélations entre variables. Nous appliquons donc une méthode de sélection de variables de type *filter*, en utilisant `SelectKBest` avec le score `f_regression`.

### Pipeline de modélisation

L’ensemble des modèles prédictifs est implémenté au sein d’un **pipeline scikit-learn** commun, comprenant :

  <img src="../figure/score_model.png" alt="Performance des modèles" style="width: 350px;"/>
</center>

- La **régression linéaire** et la **régression Ridge** présentent des performances très similaires, avec un R² d’environ **0,93** et des valeurs proches de RMSE et de MAE.  
- Le léger avantage de la régression Ridge par rapport à l’OLS suggère qu’après la sélection des variables, la multicolinéarité est limitée et que la structure linéaire est déjà bien adaptée aux données.  
- En revanche, l’**arbre de décision** affiche un R² nettement plus faible (environ **0,70**) et des erreurs de prévision plus élevées.

Sur la base de ces résultats, une **régression Ridge** utilisant les 10 variables les plus informatives apparaît comme un compromis pertinent entre **précision prédictive** et **interprétabilité** pour la prévision de la demande quotidienne d’électricité régionale.
